```python
import os
import torch
from transformers import pipeline
from tqdm import tqdm

# Cấu hình
model_id = "vinai/PhoWhisper-medium"
device = "cuda" if torch.cuda.is_available() else "cpu"
audio_dir = "output/freevc_ecapa_nonpre"
output_path = "output/freevc_ecapa_nonpre/stt_phowhisper.txt"

print(f"Loading {model_id} on {device}...")
# pipeline tự động xử lý sample rate
transcriber = pipeline("automatic-speech-recognition", model=model_id, device=device)

wav_files = [f for f in os.listdir(audio_dir) if f.endswith(".wav")]
wav_files.sort()

results = []
print("Running STT...")

for fname in tqdm(wav_files):
    path = os.path.join(audio_dir, fname)
    try:
        # PhoWhisper cực kỳ mạnh, có thể xử lý file nhiễu
        out = transcriber(path)
        text = out['text']
        print(f"{fname}: {text}")
        results.append(f"{fname}|{text}")
    except Exception as e:
        print(f"Error {fname}: {e}")
        results.append(f"{fname}|")

with open(output_path, "w", encoding="utf-8") as f:
    f.write("\n".join(results))
print("Done STT.")
```

```python
from jiwer import wer, cer
import re
from collections import defaultdict

def load_transcript(path):
    data = {}
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            if '|' not in line:
                continue
            fname, text = line.strip().split('|', 1)
            data[fname.strip()] = text.strip()
    return data

def extract_src_name(fname):
    """
    Chuyển:
    F_B30_to_F_B30__S1-S2.wav  ->  F_S1_B30.wav

    Quy tắc:
    - group(3) = target gender (F/M)
    - group(5) = speaker trước dấu '-' (S1/U1...)
    - group(4) = target age (B30, N60...)
    """
    # Thử pattern chuẩn: <srcG>_<srcAge>_to_<tgtG>_<tgtAge>__<spkA>-<spkB>.wav
    m = re.match(r'^([^_]+)_([^_]+)_to_([^_]+)_([^_]+)__([A-Z]\d+)(?:-[A-Z]\d+)?', fname)
    if not m:
        # fallback: nếu có định dạng khác, cố gắng lấy phần sau 'to_' rồi tách
        m2 = re.search(r'to_([^_]+)_([^_]+)__([A-Z]\d+)', fname)
        if not m2:
            return None
        tgt_gender = m2.group(1)
        tgt_age = m2.group(2)
        spk = m2.group(3)
        return f"{tgt_gender}_{spk}_{tgt_age}.wav"
    tgt_gender = m.group(3)
    tgt_age = m.group(4)
    spk = m.group(5)
    return f"{tgt_gender}_{spk}_{tgt_age}.wav"

# Load transcript
gt = load_transcript("output/output/freevc_pretrained/GT.txt")
base = load_transcript("output/output/freevc_pretrained/stt_phowhisper.txt")

# Gom nhóm hypotheses theo src (theo cách trích mới)
groups = defaultdict(list)

for base_fname, hyp_text in base.items():
    src = extract_src_name(base_fname)
    if src is None:
        continue
    if src in gt:
        groups[src].append(hyp_text)

# Tính WER/CER trung bình cho mỗi src
group_wers = []
group_cers = []

for src, hyps in groups.items():
    ref = gt[src]
    wers = [wer(ref, hyp) for hyp in hyps]
    cers = [cer(ref, hyp) for hyp in hyps]
    group_wers.append(sum(wers) / len(wers))
    group_cers.append(sum(cers) / len(cers))

# Mean toàn bộ
WER = sum(group_wers) / len(group_wers) if group_wers else 0
CER = sum(group_cers) / len(group_cers) if group_cers else 0

print("Number of src groups used:", len(group_wers))
print("WER:", WER)
print("CER:", CER)
```

```python
from jiwer import wer, cer
import re
import string
from collections import defaultdict

# Hàm chuẩn hóa text (BẮT BUỘC để tính WER chính xác)
def normalize_text(text):
    if not text: return ""
    text = text.lower()
    # Xóa dấu câu (chấm, phẩy, hỏi...)
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Xóa khoảng trắng thừa
    return " ".join(text.split())

def load_transcript(path):
    data = {}
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            if '|' not in line: continue
            fname, text = line.strip().split('|', 1)
            # Chuẩn hóa ngay khi load
            data[fname.strip()] = normalize_text(text)
    return data

def map_converted_to_ground_truth(fname):
    """
    Input:  F_B30_to_F_B60__S1-S2.wav (Source: S1_B30, Target: S2_B60)
    Output: F_S1_B30.wav (File chứa nội dung gốc)
    
    Quy tắc tên file VC của bạn: 
    SrcG_SrcAge_to_TgtG_TgtAge__SrcSpk-TgtSpk.wav
    """
    # Regex bắt đúng các nhóm
    # Group 1: Src Gender (F)
    # Group 2: Src Age (B30)
    # Group 5: Src Speaker (S1) (Phần trước dấu gạch ngang ở đuôi)
    pattern = r'^([A-Z])_([A-Z]\d+)_to_([A-Z])_([A-Z]\d+)__([A-Z]\d+)-([A-Z]\d+)\.wav$'
    m = re.match(pattern, fname)
    
    if not m:
        # Fallback cho trường hợp U1-U2 hoặc tên khác
        # Cố gắng tìm pattern chung: Lấy phần đầu tiên trước chữ "_to_"
        try:
            parts = fname.split('_to_')
            src_part = parts[0] # F_B30
            
            suffix = parts[1].split('__')[1] # S1-S2.wav
            src_spk = suffix.split('-')[0]   # S1
            
            src_gender = src_part.split('_')[0]
            src_age = src_part.split('_')[1]
            
            return f"{src_gender}_{src_spk}_{src_age}.wav"
        except:
            return None

    src_gender = m.group(1)
    src_age = m.group(2)
    src_spk = m.group(5) # Lấy Source Speaker!
    
    # Tên file GT tương ứng
    return f"{src_gender}_{src_spk}_{src_age}.wav"

# --- MAIN CALCULATION ---

# 1. Load data
gt_path = "output/freevc_ecapa_nonpre/GT.txt" # Đảm bảo file này chứa nội dung của SOURCE
hyp_path = "output/freevc_ecapa_nonpre/stt_phowhisper.txt"

gt_data = load_transcript(gt_path)
hyp_data = load_transcript(hyp_path)

total_wer = 0
total_cer = 0
count = 0

print(f"{'Converted File':<40} | {'Ref (Source)':<20} | WER")
print("-" * 80)

for hyp_file, hyp_text in hyp_data.items():
    # Tìm file gốc tương ứng
    gt_file = map_converted_to_ground_truth(hyp_file)
    
    if gt_file and gt_file in gt_data:
        ref_text = gt_data[gt_file]
        
        # Bỏ qua nếu GT rỗng hoặc STT rỗng
        if not ref_text or not hyp_text:
            continue
            
        error_w = wer(ref_text, hyp_text)
        error_c = cer(ref_text, hyp_text)
        
        total_wer += error_w
        total_cer += error_c
        count += 1
        
        if count <= 5: # In thử 5 dòng đầu check mapping
            print(f"{hyp_file[:35]}... | {gt_file} | {error_w:.2f}")

if count > 0:
    avg_wer = total_wer / count
    avg_cer = total_cer / count
    print("\n" + "="*30)
    print(f"Evaluated {count} pairs.")
    print(f"Average WER: {avg_wer:.4f}")
    print(f"Average CER: {avg_cer:.4f}")
    print("="*30)
else:
    print("No matching ground truth files found! Check your filename mapping logic.")
```

```python
import os
import torch
from transformers import pipeline
from tqdm import tqdm

# Cấu hình
model_id = "vinai/PhoWhisper-medium"
device = "cuda" if torch.cuda.is_available() else "cpu"
audio_dir = "CER_WER"
output_path = "CER_WER/stt_phowhisper.txt"

print(f"Loading {model_id} on {device}...")
# pipeline tự động xử lý sample rate
transcriber = pipeline("automatic-speech-recognition", model=model_id, device=device)

wav_files = [f for f in os.listdir(audio_dir) if f.endswith(".wav")]
wav_files.sort()

results = []
print("Running STT...")

for fname in tqdm(wav_files):
    path = os.path.join(audio_dir, fname)
    try:
        # PhoWhisper cực kỳ mạnh, có thể xử lý file nhiễu
        out = transcriber(path)
        text = out['text']
        print(f"{fname}: {text}")
        results.append(f"{fname}|{text}")
    except Exception as e:
        print(f"Error {fname}: {e}")
        results.append(f"{fname}|")

with open(output_path, "w", encoding="utf-8") as f:
    f.write("\n".join(results))
print("Done STT.")
```
